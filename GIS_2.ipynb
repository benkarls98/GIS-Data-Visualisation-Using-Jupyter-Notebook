```python
# A function to load any layer
from shapely import wkt
from shapely.validation import make_valid

def load_layer(name):
    d = pd.read_csv(z.open(name), sep=None, engine="python", encoding="utf-8")
    geo = gpd.GeoDataFrame(d, geometry=d["WKT"].apply(wkt.loads), crs="EPSG:4326")
    geo = geo.drop(columns="WKT").to_crs(g.crs)   # same metric CRS as vegetation
    geo["geometry"] = geo.geometry.apply(make_valid)
    return geo

water = load_layer("data/water-polygon.csv")
print(len(water), "water polygons")
```

```python
#How much vegetation is within 500m of water?
# How much vegetation is within 500 m of water?
water_buf = water.geometry.buffer(500)
if hasattr(water_buf, "union_all"):
    water_zone = water_buf.union_all()
else:
    water_zone = water_buf.unary_union

veg_total = veg.area
veg_near_water = veg.intersection(water_zone).area
pct = veg_near_water / veg_total * 100

print(f"Total vegetation: {veg_total / 10_000:,.1f} ha")
print(f"Within 500 m of water: {veg_near_water / 10_000:,.1f} ha ({pct:.1f}%)")
```

```python
#Load the Buildings
buildings = load_layer("data/building-polygon.csv")
print(len(buildings), "buildings")
```

```python
# What share of the buildings sit next to green space?
veg_zone = gpd.GeoDataFrame(geometry=[veg.buffer(100)], crs=g.crs)   # within 50 m of vegetation

near = gpd.sjoin(buildings, veg_zone, predicate="intersects", how="inner")
share = len(near) / len(buildings) * 100

print(f"{len(near):,} of {len(buildings):,} buildings ({share:.1f}%) are within 50 m of vegetation")
```

```python
# Compare the distances in a chart
dists = [25, 50, 100, 250, 500]
shares = []

for d in dists:
    zone = gpd.GeoDataFrame(geometry=[veg.buffer(d)], crs=g.crs)
    n = len(gpd.sjoin(buildings, zone, predicate="intersects", how="inner"))
    shares.append(n / len(buildings) * 100)

plt.figure(figsize=(7, 4))
plt.plot(dists, shares, marker="o")
plt.xlabel("Distance to vegetation (m)")
plt.ylabel("Buildings within distance (%)")
plt.title("Access to green space")
plt.grid(True)
plt.show()
```

```python
# ===== Green Space Dashboard =====
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from matplotlib.patches import FancyBboxPatch

# ---- 1. Numbers for the tiles ----
veg_ha        = veg_total / 10_000
n_patches     = len(gdf)
pct_near_wat  = veg_near_water / veg_total * 100
pct_bld_green = share                      # % of buildings within 50 m of vegetation

# ---- 2. Colours ----
BG, CARD, TXT, MUTED = "#0f1b16", "#17271f", "#e8f1ec", "#8fa89b"
GREEN, LIME, BLUE, RED = "#2ecc71", "#a8e063", "#3498db", "#e74c3c"

plt.rcParams.update({"text.color": TXT, "axes.labelcolor": MUTED,
                     "xtick.color": MUTED, "ytick.color": MUTED,
                     "axes.edgecolor": CARD, "font.size": 10})

fig = plt.figure(figsize=(16, 10), facecolor=BG)
gs = GridSpec(3, 4, figure=fig, height_ratios=[0.55, 1, 1], hspace=0.45, wspace=0.3)

fig.suptitle("Green Space Dashboard", fontsize=24, fontweight="bold",
             color=TXT, x=0.075, y=0.99, ha="left")

# ---- 3. KPI tiles ----
tiles = [("Total vegetation", f"{veg_ha:,.0f} ha", GREEN),
         ("Vegetation patches", f"{n_patches:,}", LIME),
         ("Vegetation within 500 m of water", f"{pct_near_wat:.0f}%", BLUE),
         ("Buildings within 50 m of green", f"{pct_bld_green:.0f}%", RED)]
for i, (label, value, colour) in enumerate(tiles):
    ax = fig.add_subplot(gs[0, i])
    ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.axis("off")
    ax.add_patch(FancyBboxPatch((0, 0), 1, 1, boxstyle="round,pad=0,rounding_size=0.08",
                                fc=CARD, ec="none", transform=ax.transAxes))
    ax.add_patch(FancyBboxPatch((0, 0.05), 0.025, 0.9, boxstyle="round,pad=0,rounding_size=0.01",
                                fc=colour, ec="none"))
    ax.text(0.1, 0.62, value, fontsize=26, fontweight="bold", color=colour, va="center")
    ax.text(0.1, 0.22, label, fontsize=10, color=MUTED, va="center")

# ---- 4. Coverage map ----
ax = fig.add_subplot(gs[1:, 0:2])
grid.plot(column="veg_pct", cmap="YlGn", ax=ax, legend=True,
          legend_kwds={"label": "Vegetation cover (%)", "shrink": 0.6})
ax.set_facecolor(CARD); ax.set_axis_off()
ax.set_title("Vegetation cover per 1 km cell", loc="left", color=TXT, fontsize=13, pad=10)

# ---- 5. Area by type ----
ax = fig.add_subplot(gs[1, 2])
by_type = gdf.groupby("type")["area_ha"].sum().sort_values().tail(6)
ax.barh(by_type.index.astype(str), by_type.values, color=GREEN)
ax.set_facecolor(CARD)
ax.set_title("Area by type (ha)", loc="left", color=TXT, fontsize=13)
for s in ax.spines.values(): s.set_visible(False)

# ---- 6. Patch size distribution ----
ax = fig.add_subplot(gs[1, 3])
vals = gdf["area_ha"][gdf["area_ha"] > 0]
ax.hist(np.log10(vals), bins=30, color=LIME)
ax.set_facecolor(CARD)
ax.set_xlabel("Patch area (log10 ha)")
ax.set_title("Patch size distribution", loc="left", color=TXT, fontsize=13)
for s in ax.spines.values(): s.set_visible(False)

# ---- 7. Access to green space ----
ax = fig.add_subplot(gs[2, 2:4])
ax.plot(dists, shares, marker="o", color=RED, lw=2.5)
ax.fill_between(dists, shares, color=RED, alpha=0.15)
ax.set_facecolor(CARD)
ax.set_xlabel("Distance to vegetation (m)")
ax.set_ylabel("Buildings within distance (%)")
ax.set_ylim(0, 105)
ax.set_title("Access to green space", loc="left", color=TXT, fontsize=13)
ax.grid(axis="y", color=BG, alpha=0.6)
for s in ax.spines.values(): s.set_visible(False)

plt.show()
fig.savefig("green_space_dashboard.png", dpi=200, bbox_inches="tight", facecolor=BG)
```

```python
# ===== Interactive Green Space Dashboard (Plotly) =====
# First time only:  !pip install plotly
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

BUILD_DIST = 50     # the distance (m) you used for the buildings "share" number

# ---- Colours ----
BG, CARD, TXT, MUTED = "#0f1b16", "#17271f", "#e8f1ec", "#8fa89b"
GREEN, LIME, BLUE, RED = "#2ecc71", "#a8e063", "#3498db", "#e74c3c"

# ---- Data prepared from your earlier results ----
veg_ha       = veg_total / 10_000
pct_water    = veg_near_water / veg_total * 100

by_type = gdf.groupby("type")["area_ha"].sum().sort_values().tail(8)
type_share = by_type / gdf["area_ha"].sum() * 100

vals = gdf["area_ha"][gdf["area_ha"] > 0]
counts, edges = np.histogram(np.log10(vals), bins=30)
centers = (edges[:-1] + edges[1:]) / 2
lo_ha, hi_ha = 10 ** edges[:-1], 10 ** edges[1:]

# coverage grid -> regular matrix for a heatmap
cx = grid.geometry.centroid.x.values
cy = grid.geometry.centroid.y.values
step = grid.geometry.bounds.iloc[0].pipe(lambda b: b["maxx"] - b["minx"])
ix = np.round((cx - cx.min()) / step).astype(int)
iy = np.round((cy - cy.min()) / step).astype(int)
z = np.full((iy.max() + 1, ix.max() + 1), np.nan)
z[iy, ix] = grid["veg_pct"].values
xs_km = (cx.min() + np.arange(z.shape[1]) * step) / 1000
ys_km = (cy.min() + np.arange(z.shape[0]) * step) / 1000

# ---- Layout ----
ind = {"type": "indicator"}
fig = make_subplots(
    rows=3, cols=4,
    specs=[[ind, ind, ind, ind],
           [{"rowspan": 2, "colspan": 2}, None, {}, {}],
           [None, None, {"colspan": 2}, None]],
    row_heights=[0.2, 0.4, 0.4], vertical_spacing=0.12, horizontal_spacing=0.09,
    subplot_titles=("", "", "", "", "Vegetation cover per cell (%)", "Area by type (ha)",
                    "Patch size distribution", "Access to green space"))

# ---- KPI tiles ----
tiles = [("Total vegetation", veg_ha, " ha", ",.0f", GREEN),
         ("Vegetation patches", len(gdf), "", ",d", LIME),
         ("Vegetation within 500 m of water", pct_water, "%", ".0f", BLUE),
         (f"Buildings within {BUILD_DIST} m of green", share, "%", ".0f", RED)]
for i, (label, value, suffix, fmt, colour) in enumerate(tiles, start=1):
    fig.add_trace(go.Indicator(
        mode="number", value=value,
        number={"suffix": suffix, "valueformat": fmt, "font": {"color": colour, "size": 44}},
        title={"text": label, "font": {"size": 13, "color": MUTED}}), row=1, col=i)

# ---- Coverage map ----
fig.add_trace(go.Heatmap(
    z=z, x=xs_km, y=ys_km, colorscale="YlGn", zmin=0, zmax=100,
    colorbar={"orientation": "h", "title": {"text": "Vegetation cover (%)", "side": "top"},
              "len": 0.4, "x": 0.2, "y": -0.12, "thickness": 12},
    hovertemplate="<b>%{z:.0f}% vegetation</b><br>x: %{x:.1f} km, y: %{y:.1f} km<extra></extra>",
    hoverongaps=False), row=2, col=1)
heat_idx = len(fig.data) - 1
fig.update_yaxes(scaleanchor=fig.data[heat_idx].xaxis, row=2, col=1)

# ---- Area by type ----
palette = [GREEN, LIME, "#1abc9c", "#27ae60", "#82e0aa", "#52be80", "#7dcea0", "#58d68d"]
fig.add_trace(go.Bar(
    x=by_type.values, y=by_type.index.astype(str), orientation="h",
    marker_color=palette[:len(by_type)], customdata=type_share.values,
    hovertemplate="<b>%{y}</b><br>%{x:,.0f} ha<br>%{customdata:.1f}% of all vegetation<extra></extra>"),
    row=2, col=3)

# ---- Patch size distribution ----
fig.add_trace(go.Bar(
    x=centers, y=counts, marker_color=LIME, customdata=np.c_[lo_ha, hi_ha],
    hovertemplate="<b>%{y} patches</b><br>%{customdata[0]:,.1f} to %{customdata[1]:,.1f} ha<extra></extra>"),
    row=2, col=4)
nice = [1, 2, 5, 10, 20, 50, 100, 200, 500, 1000, 2000, 5000, 10000, 20000, 50000]
nice = [v for v in nice if 10 ** edges[0] <= v <= 10 ** edges[-1]] or [round(10 ** edges[0], 1)]
fig.update_xaxes(tickvals=[np.log10(v) for v in nice], ticktext=[f"{v:,}" for v in nice],
                 title_text="Patch area (ha, log scale)", row=2, col=4)

# ---- Access curve ----
fig.add_trace(go.Scatter(
    x=dists, y=shares, mode="lines+markers", fill="tozeroy",
    line={"color": RED, "width": 3}, marker={"size": 10},
    fillcolor="rgba(231,76,60,0.15)",
    hovertemplate="<b>%{y:.1f}% of buildings</b><br>within %{x} m of vegetation<extra></extra>"),
    row=3, col=3)
fig.update_xaxes(title_text="Distance to vegetation (m)", row=3, col=3)
fig.update_yaxes(title_text="Buildings (%)", range=[0, 105], row=3, col=3)

# ---- Colour-scale switcher for the map ----
buttons = [dict(label=name, method="restyle", args=[{"colorscale": name}, [heat_idx]])
           for name in ["YlGn", "Greens", "Viridis", "Cividis"]]

fig.update_layout(
    title={"text": "<b>Green Space Dashboard</b>", "x": 0.02, "font": {"size": 26}},
    paper_bgcolor=BG, plot_bgcolor=CARD, font={"color": TXT},
    height=820, showlegend=False, margin={"l": 40, "r": 30, "t": 90, "b": 90},
    hoverlabel={"bgcolor": CARD, "font_size": 13},
    updatemenus=[dict(type="dropdown", buttons=buttons, x=0.35, y=1.12, xanchor="left",
                      yanchor="top", bgcolor=CARD, font={"color": TXT})])
fig.update_yaxes(automargin=True, row=2, col=3)
fig.update_xaxes(gridcolor=BG, zeroline=False)
fig.update_yaxes(gridcolor=BG, zeroline=False)
fig.update_annotations(font={"size": 13, "color": TXT})

fig.show()
fig.write_html("green_space_dashboard.html")   # shareable file, opens in any browser
```
